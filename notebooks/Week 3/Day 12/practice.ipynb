```python
class DivisionByZeroError (Exception):
    pass
def safe_divide(a, b):
    if b == 0:
       raise DivisionByZeroError("cannot divide by zero")
    return a/b

try:
    result = safe_divide(1, 0)
    print("Result:", result)

except DivisionByZeroError as e:
    print("Friendly message:", e)

finally:
    print("done") 


        
```
