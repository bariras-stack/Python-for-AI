```python
class Shape:
    def area(self):
        return 0

class rectangle(Shape):
    def __init__(self, height, width):
        self.height = height
        self.width = width

    def area(self):
        return self.width * self.height

class circle(Shape):
    def __init__(self, radius):
        self.radius =  radius

    def area(self):
        return 3.14 * self.radius ** 2

shapes = [circle(2), rectangle(2, 5)]
for shape in shapes: 
    print (shape.area())
```

```python
class book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author =  author
        self.isbn =  isbn

    def __str__(self):
        return f"{self.title}'s author is {self.author}"

    def __eq__(self, other):
        return self.isbn == other.isbn

book1 = book("Harry pother", "JK rowling", "123")
book2 = book("kite runner", "khaled", "123")

print(book1)
print(book1==book2)
        
```
