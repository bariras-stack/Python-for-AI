```python
def shout(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result.upper()
    return wrapper


@shout
def greet(name):
    return f"hello {name}"


print(greet("Barira"))
```

```python
def shout(func):
    def inner(name):
        return func(name).upper()
    return inner

name = input ("enter name ")

@shout
def greet(name):
    return f"hello {name}!"

print (greet(name))
```

```python
def count_up_to(n):
    i = 1
    while i <= n:
        yield i          # hand out one value, then pause right here
        i += 1
        
        # yield i    

for number in count_up_to(5):
    print(number)
```
