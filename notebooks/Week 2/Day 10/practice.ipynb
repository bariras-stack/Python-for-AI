```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) // len(self.marks)

    def grade(self):
        
        avg = self.average()

        if avg >= 80:
            return ("A+")
        elif avg >= 70:
            return ("B")
        elif avg >= 60:
            return("C")
        else:
            return("fail")
        
student = Student ("Barira", [85, 95, 70])   
print("name: ", student.name) 
print("average:", student.average())
print("grade:", student.grade())


```
