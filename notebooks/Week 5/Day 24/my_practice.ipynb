````python
from openai import OpenAI
from dotenv import load_dotenv

import os
import json

load_dotenv()
````

````python
llm_client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPEN_ROUTER_KEY"),
)

MODEL = "nvidia/nemotron-3.5-lightning:free"   
````

````python
messages = [
    {
      "role": "system",  # User | Assistant | System
      "content": """
        I have a tea shop, and you're my assistant and your job it classify customer messages into a
        specific format which i'll provide you (It should be in JSON).

        FOR "label" KEY = complaint | praise | question
        FOR "reason" KEY = give me in 4 to 5 words

        RESPONSE FORMAT:
        {"label": "complaint", "reason": "cold chai, late"}
      """
    }
]
````

````python
def parse_json(text):
    text = text.strip().removeprefix("```json").removeprefix("```").removesuffix("```").strip()   # remove fences
    return json.loads(text) 
````

````python
def ask_assistant(user_query):
    try:

        messages.append({
            "role": "user",
            "content": user_query
        })
        
        completion = llm_client.chat.completions.create(
          model=MODEL,
          messages=messages
        )
        reply = completion.choices[0].message.content
        reply = json.loads(reply)

        messages.append({
            "role": "assistant",
            "content": reply
        })

        print(type(reply), reply)

        return reply
    except Exception as e:
        print(f"Error occured: {e}")
````

````python
ask_assistant("The Chai was karak, amazing!!! and all good")
````
